## Part 4: Loops

A **for loop** repeats code for every element:

```python
for color in ["red", "green", "blue"]:
    print(color)

for i in range(1, 4):   # 1, 2, 3  (the end is NOT included)
    print(i)
```

A **while loop** repeats as long as a condition is `True`:

```python
n = 3
while n > 0:
    print(n)
    n = n - 1
```

### Exercise 4.1 – Sum with a loop

Calculate the sum of `numbers` **with a for loop** (not with `sum()`).

Tip: start with `total = 0` and add every number.

````python
numbers = [4, 8, 15, 16, 23, 42]
total = 0

for number in numbers:
    total = total + number

print(total)
````

````python
# Check your answer
check(total == 108, "inside the loop: total = total + number")
````

### Exercise 4.2 – Loop + if

Create a list `even_numbers` with all even numbers from 1 to 20.

Tip: `number % 2 == 0` is `True` if a number is even (`%` is the remainder of a division).

````python
even_numbers = []

for number in range(1, 21):
    if number % 2 == 0:
        even_numbers.append(number)

print(even_numbers)
````

````python
# Check your answer
check(even_numbers == list(range(2, 21, 2)),
      "range(1, 21) goes from 1 to 20")
````

### Exercise 4.3 – While loop

Use a **while** loop to create the list `countdown = [5, 4, 3, 2, 1]`.

````python
countdown = []
n = 5

while n > 0:
    countdown.append(n)
    n = n - 1

print(countdown)
````

````python
# Check your answer
check(countdown == [5, 4, 3, 2, 1],
      "do not forget n = n - 1, otherwise the loop never ends")
````

## Part 5: Dictionaries

A **dictionary** stores **key → value** pairs. It is like one row of a
database table!

```python
product = {"name": "Chai", "price": 18.0, "category": "Beverages"}
print(product["price"])      # 18.0
product["stock"] = 39        # add a new key

for key, value in product.items():
    print(key, "=", value)
```

### Exercise 5.1 – Using a dictionary

1. Save the country of the customer in `country`.
2. Add a new key `"city"` with the value `"Berlin"`.

````python
customer = {"name": "Alfreds Futterkiste", "country": "Germany"}

country = customer["country"]
customer["city"] = "Berlin"

print(customer)
````

````python
# Check your answer
check(country == "Germany" and customer.get("city") == "Berlin",
      'use customer["country"] and customer["city"] = ...')
````

### Exercise 5.2 – Counting with a dictionary

Count how often each country appears in `countries`.
The result should be `{"Germany": 3, "France": 2, "Spain": 1}`.

Tip: `if country in counts:` checks if a key already exists.

(Later you will do the same in SQL with `GROUP BY`!)

````python
countries = ["Germany", "France", "Germany", "Spain", "France", "Germany"]
counts = {}

for country in countries:
    if country in counts:
        counts[country] = counts[country] + 1
    else:
        counts[country] = 1

print(counts)
````

````python
# Check your answer
check(counts == {"Germany": 3, "France": 2, "Spain": 1},
      "new country: counts[country] = 1, else add 1")
````

A **function** is reusable code with a name. It gets input (**parameters**)

### Exercise 6.3 – Function with a loop

Write `count_long_words(words, min_length)` that returns how many words have **at least** `min_length` letters.

````python
def count_long_words(words, min_length):
    count = 0
    for word in words:
        if len(word) >= min_length:
            count = count + 1
    return count

print(count_long_words(["sql", "python", "data", "database"], 5))
````

````python
# Check your answer
check(count_long_words(["sql", "python", "data", "database"], 5) == 2
      and count_long_words(["a", "bb"], 1) == 2
      and count_long_words([], 3) == 0,
      "count with a loop and len(word)")
````

## Part 7: Mini project – a small "database" in Python

A list of dictionaries looks just like a database table: each dictionary is
a **row**, each key is a **column**. This is the bridge to SQL!

### Exercise 7.1 – Filter and calculate

Using the list `products`:

1. `expensive` – a list with the **names** of all products with a price over 20.
2. `stock_value` – the total value of all products (price × stock for each product, added up).

In SQL, part 1 would be: `SELECT name FROM products WHERE price > 20`

````python
products = [
    {"name": "Chai", "price": 18.0, "stock": 39},
    {"name": "Chang", "price": 19.0, "stock": 17},
    {"name": "Mishi Kobe Niku", "price": 97.0, "stock": 29},
    {"name": "Ikura", "price": 31.0, "stock": 31},
    {"name": "Konbu", "price": 6.0, "stock": 24},
]

expensive = []
stock_value = 0

for product in products:
    if product["price"] > 20:
        expensive.append(product["name"])
    stock_value = stock_value + product["price"] * product["stock"]

print(expensive)
print(stock_value)
````

````python
# Check your answer
check(expensive == ["Mishi Kobe Niku", "Ikura"] and stock_value == 4943,
      "loop over products and use product[\"price\"]")
````
